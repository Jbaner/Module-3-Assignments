> The raw Zillow dataset required substantial cleaning before any modeling could begin. It contained many of the challenges typical in real-world real estate data: extensive missing values, redundant features, inconsistent data types, and several variables that were unusable for prediction. The goal of preprocessing was to transform this messy dataset into a reliable, fully numeric feature matrix suitable for model comparison, while preserving the information most relevant to predicting tax values.

> A major issue in the raw data was the extremely high percentage of missing values in many columns. Several features—such as basementsqft, architecturalstyletypeid, finishedsquarefeet13, decktypeid, storytypeid, and multiple pool-related variables had over 95% missing values. These features were removed because they provided little usable information and would introduce noise rather than improve predictions. Identifier variables like parcelid were also dropped because they uniquely identify properties but do not describe them.

> For the remaining features, missing values were handled using type-appropriate imputation. Numeric variables were imputed using the median to reduce the influence of outliers. Ordinal variables, such as building quality ratings, were imputed with median values and encoded using an ordinal encoder. Categorical variables were imputed with "Unknown" and then encoded using either frequency encoding (for high-cardinality columns like ZIP code and neighborhood) or one-hot encoding (for low-cardinality features like heating and cooling types). Boolean flags were mapped to numeric values to ensure full compatibility with tree-based models.

> Several preprocessing alternatives were considered. For example, one option was to one-hot encode all categorical variables, but this would have created thousands of sparse columns due to high-cardinality geographic features. Frequency encoding was chosen instead because it preserves signal while keeping the feature space manageable. Another alternative was to scale all features, but scaling was retained only for models that require it (Ridge and Linear Regression). Tree-based models were trained on unscaled data because scaling does not improve their performance.

> In the final analysis, the retained preprocessing steps were those that produced a fully numeric, compact, and information-rich dataset: dropping unusable features, imputing missing values, encoding categorical variables appropriately, and ensuring consistent transformations across training and test sets. Procedures that added complexity without improving performance—such as keeping high-missing columns or applying universal one-hot encoding—were rejected to maintain model stability and interpretability.
